## 4. Modelo de clasificación\nSe utiliza un árbol de decisión para clasificar la demanda como Alta o Baja.

```python
X_clf = df[['Precio', 'Visitas', 'Publicidad', 'Descuento']]\ny_clf = df['Demanda']\n\nX_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(\n    X_clf, y_clf, test_size=0.25, random_state=42, stratify=y_clf\n)\n\nmodelo_clasificacion = DecisionTreeClassifier(max_depth=3, random_state=42)\nmodelo_clasificacion.fit(X_train_clf, y_train_clf)\npred_clf = modelo_clasificacion.predict(X_test_clf)\n\nexactitud = accuracy_score(y_test_clf, pred_clf)\nprint(f'Exactitud: {exactitud:.2f}')\nprint('\\nMatriz de confusión:')\nmatriz = confusion_matrix(y_test_clf, pred_clf, labels=['Baja', 'Alta'])\nprint(matriz)\nprint('\\nReporte de clasificación:')\nprint(classification_report(y_test_clf, pred_clf, zero_division=0))\n
```

```python
ConfusionMatrixDisplay(\n    confusion_matrix=matriz,\n    display_labels=['Baja', 'Alta']\n).plot()\nplt.title('Matriz de confusión')\nplt.show()\n\nplt.figure(figsize=(12, 6))\nplot_tree(\n    modelo_clasificacion,\n    feature_names=X_clf.columns,\n    class_names=modelo_clasificacion.classes_,\n    filled=True,\n    rounded=True\n)\nplt.title('Árbol de decisión para demanda')\nplt.show()\n
```

### Análisis de clasificación\n- Se clasifican correctamente **5 de 5** registros de prueba.\n- Se identifican correctamente **4 de 4** casos de demanda Alta.\n- Un falso positivo sería predecir demanda Alta cuando la clase real es Baja.\n- Ese error podría producir sobreinventario o gasto publicitario innecesario.\n- La exactitud de 1.00 no debe generalizarse: la muestra es ficticia y la prueba contiene solo cinco registros.
